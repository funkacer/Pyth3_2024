```python
def sum2dArr(arr):
    result=0.0
    M,N=arr.shape
    for i in range(M):
        for j in range(N):
            result+=arr[i,j]
    return arr
```

```python
import time
```

```python
start=time.perf_counter()
for _ in range(10):
    sum2dArr(bigA)
print("Celkovy cas : ",time.perf_counter()-start)
```

```python
@jit
def sum2dArr_JIT(arr):
    result=0.0
    M,N=arr.shape
    for i in range(M):
        for j in range(N):
            result+=arr[i,j]
    return arr
```

```python
start=time.perf_counter()
for _ in range(10):
    sum2dArr_JIT(bigA)
print("Celkovy cas : ",time.perf_counter()-start)
```

# Lotka-Volterra

\begin{align}
\frac{dx}{dt}&=\alpha x - \beta x y \\
\frac{dy}{dt}&=\delta x y -\gamma y
\end{align}

```python
from scipy.integrate import odeint
import pylab as p
%matplotlib inline
```

```python
alfa,beta,gama,delta=1.0,0.1,1.0,0.075
Zajace,Vlci=15,5
t=np.linspace(1,15,100)
```

```python
def lv(init,t):
    x,y=init[0],init[1]
    dx_dt=alfa*x-beta*x*y
    dy_dt=delta*x*y-gama*y
    return np.array([dx_dt,dy_dt])
```

```python
X0=np.array([Zajace,Vlci])
X=odeint(lv,X0,t)
z,v=X.T
```

```python
z
```

```python
v
```

```python
p.plot(t,z,'b-',label='Zajace')
p.plot(t,v,'r-',label='Vlci')
p.legend()
```

```python
@interact(Zajace=(15,100,1),Vlci=(5,25,1))
def simulation(Zajace,Vlci):
    X0=np.array([Zajace,Vlci])
    X=odeint(lv,X0,t)
    z,v=X.T
    p.plot(t,z,'b-',label='Zajace')
    p.plot(t,v,'r-',label='Vlci')
    p.legend()
```

```python
class display(object):
    template = """<div style="float: left; padding: 10px;">
    <p style = 'font-family:"Cou  rier New", Courier, monospace'>{0}</p>{1}
    </div>"""
    
    def __init__(self, *a):
        self.args = a
    
    def _repr_html_(self):
        return '\n'.join(self.template.format(a,eval(a)._repr_html_())
                        for a in self.args)
    def __repr__(self):
        return '\n\n'.join(a+'\n'+repr(eval(a))
                          for a in self.args)
```
