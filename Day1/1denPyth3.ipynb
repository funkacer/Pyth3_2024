```python
def getParams(f,count=10):
    lineCount, wordCount=0,0
    result=defaultdict(int)
    subor=open(f,'r')
    for line in subor.readlines():
        lineCount+=1
        line=line.strip().split()
        wordCount+=len(line)
        for word in line:
            result[word]+=1
    subor.close() 
    """
    top={}
    for word,freq in sorted(result.items(),key=lambda x:x[1],reverse=True)[:count]:
        top.update({word:freq})
    """
    top={word:freq for word,freq in sorted(result.items(),key=lambda x:x[1],reverse=True)[:count]}
    return (lineCount,wordCount,top)
```

```python
def getPie(nazov,x):
    fig = plt.figure(1,figsize=(6,6),dpi=100)
    ax=fig.add_axes([0.6,0.1,0.6,0.6])
    key=list(x.keys())
    value=list(x.values())
    expl=[0.0 for _ in range(len(x))]
    expl[0]=0.15
    ax.pie(value,labels=key,explode=expl,shadow=True,autopct="%1.2f%%")
    plt.title(nazov)
    plt.show()
```

# ORM

```python
from sqlalchemy import Column,Integer,String,create_engine
from sqlalchemy.ext.declarative import declarative_base as BASE
```

```python
conn=create_engine("mysql://root:root@10.2.120.20/tibor")
```

```python
base=BASE()
base.metadata.create_all(conn)
```

```python
%sql desc tibor.gopas
```

```python
class Gopas(base):
    __tablename__='gopas'
    id=Column(Integer,primary_key=True)
    name=Column(String)
    salary=Column(Integer)
    def __str__(self):
        return "{}, {}, {}".format(self.id,self.name,self.salary)
```

```python
from sqlalchemy.orm import sessionmaker
```

```python
session=sessionmaker(bind=conn)
s=session()
```

```python
data=s.query(Gopas).all()
for record in data:
    print(record)
```

```python
newRec=Gopas(name='tibor',salary=2400)
```

```python
s.add(newRec)
```

```python
s.commit()
```

```python
data=s.query(Gopas).all()
for record in data:
    print(record)
```
